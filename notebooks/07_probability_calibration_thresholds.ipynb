```python
from pathlib import Path
import json
import sys

cwd = Path.cwd().resolve()
ROOT = next(p for p in (cwd, *cwd.parents) if (p / "AGENTS.md").is_file() and (p / "pyproject.toml").is_file())
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from scripts.model_evaluation.contract import ModelEvaluationConfig
from scripts.model_evaluation.pipeline import build_model_evaluation, run_summary
```

```python
config = ModelEvaluationConfig(root=ROOT)
result = build_model_evaluation(config)
```

```python
summary = run_summary(result, ROOT)
print(json.dumps({"model_evaluation_run": summary}, indent=2))
```

```python
import pandas as pd
from IPython.display import display

calibration = pd.read_csv(result.bundle / "calibration_selection.csv")
thresholds = pd.read_csv(result.bundle / "threshold_selection.csv")
temporal_audit = pd.read_csv(result.bundle / "temporal_audit.csv")

display(calibration)
display(thresholds)

audit_flags = ["chronology_pass", "independence_pass", "component_available_pass", "no_reserved_pass"]
audit_view = temporal_audit[
    [
        "candidate",
        "window_id",
        "horizon",
        "selected_configuration_id",
        *audit_flags,
    ]
]
display(audit_view)
```

```python
posterior = pd.read_csv(result.bundle / "posterior_metrics.csv")
independent_checks = pd.read_csv(result.bundle / "independent_checks.csv")

window_means = (
    posterior[
        posterior["metric"].isin(
            ["average_precision", "brier", "onset_sensitivity", "false_alarm_ratio"]
        )
    ]
    .groupby(["candidate", "horizon", "metric"], dropna=False)["value"]
    .mean()
    .reset_index()
)
display(window_means)

display(
    independent_checks[
        ["candidate", "window_id", "horizon", "check", "pass"]
    ].sort_values(["candidate", "window_id", "horizon", "check"])
)

failed = independent_checks[~independent_checks["pass"].fillna(False)]
print(f"Independent checks failing rows: {failed.shape[0]}")
print("Stage 5 delivery 1 remains development evidence only; no champion and no reserved-block evaluation.")
```
